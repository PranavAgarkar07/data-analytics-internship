# Task 1 — Data Immersion & Wrangling

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/retail_store_sales.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
clean = df.dropna(subset=["Item"])
```

```python
clean["Price Per Unit"] = clean.groupby("Category")["Price Per Unit"].transform(
    lambda x: x.fillna(x.median())
)
```

```python
clean["Quantity"] = clean.groupby("Category")["Quantity"].transform(
    lambda x: x.fillna(x.median())
).round().astype(int)
```

```python
clean["Total Spent"] = clean["Price Per Unit"] * clean["Quantity"]
```

```python
clean["Discount Applied"] = clean["Discount Applied"].fillna(False)
```

```python
clean["Transaction Date"] = pd.to_datetime(clean["Transaction Date"])
```

```python
clean.isnull().sum()
clean.dtypes
```

```python
clean["Year"] = clean["Transaction Date"].dt.year
clean["Month"] = clean["Transaction Date"].dt.month
clean["Weekday"] = clean["Transaction Date"].dt.day_name()
```

```python
clean.to_csv("../data/retail_store_sales_clean.csv", index=False)
```
